# 1. Setup and Library Imports

```python
import pandas as pd
```

# 2. Load the Dataset

```python
df = pd.read_csv('vaccinations.csv')
```

```python
# Display the first 5 rows of the DataFrame to get a quick look at the data
print("\nStep 2: Dataset successfully loaded. Here are the first 5 rows:")
print(df.head())
```

# 3. Initial Data Exploration

- Before cleaning, it's crucial to understand the dataset. We will check the data types of each column and look for missing values.

```python
# Get a concise summary of the DataFrame, including data types and non-null counts
print("\n--- DataFrame Info ---")
df.info()
```

```python
# Get the count of missing (null) values for each column
print("\n--- Missing Values Per Column ---")
print(df.isnull().sum())
```

# 4. Data Cleaning and Preprocessing

- Handling Missing Values: A simple and effective strategy for this dataset is to fill any missing numerical values with `0`. This is a reasonable assumption, as missing data often implies that no new vaccinations were reported on that day.

```python
# Use the fillna() method to replace all NaN (Not a Number) values with 0
# The 'inplace=True' argument modifies the DataFrame directly
df.fillna(0, inplace=True)
```

```python
# Verify that there are no more missing values
total_missing = df.isnull().sum().sum()
print(f"\nMissing values handled. Total remaining missing values: {total_missing}")
```

- Correcting Data Types: The `date` column must be converted to a datetime format to enable chronological sorting, filtering, and analysis.

```python
# Convert the 'date' column from object to datetime format
df['date'] = pd.to_datetime(df['date'])
```

```python
# Check the data types again to confirm the conversion was successful
print("\n'date' column converted to datetime. Verifying info:")
df.info()
```

# 5. Save the Cleaned Data

- After cleaning and preprocessing, we'll save the modified DataFrame to a new CSV file. This allows us to use the cleaned data for future analysis without repeating these steps.

```python
# Define the filename for the cleaned data
cleaned_filename = 'vaccinations_cleaned.csv'

# Save the cleaned DataFrame to a new CSV file
# 'index=False' prevents pandas from writing the DataFrame index as a column
df.to_csv(cleaned_filename, index=False)
print(f"\nSuccessfully cleaned the data and saved it to '{cleaned_filename}'")
```

# 6. Conclusion

- The data is now clean, with no missing values and correct data types. It is stored in `vaccinations_cleaned.csv` and is ready for exploratory data analysis, visualization, and modeling.
